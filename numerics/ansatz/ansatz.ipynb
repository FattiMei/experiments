```python
import sympy as sym
from sympy import sqrt
from ansatz import (
    LinearAnsatz,
    AffineAnsatz,
    solve_ansatz,
)
```

```python
z = sym.Symbol('z')
n = sym.Symbol('n')


def functional_equation(ansatz, n):
    return sym.Eq(
        ansatz(n),
        ansatz(2*n) + z / sqrt(2*n)
    )
```

```python
class Ansatz:
    def __init__(self):
        self.C = sym.Symbol('C')
        self.args = (self.C,)

    def __call__(self, n, args: dict = {}):
        C = self.C

        return C / sqrt(n)


solution = solve_ansatz(functional_equation, Ansatz(), n)
solution
```

```python
class ComplicatedAnsatz:
    def __init__(self):
        self.A = sym.Symbol('A')
        self.B = sym.Symbol('B')
        self.C = sym.Symbol('C')
        self.D = sym.Symbol('D')
        self.args = (self.A, self.B, self.C, self.D)

    def __call__(self, n):
        A = self.A
        B = self.B
        C = self.C
        D = self.D

        return (A + B*sqrt(n)) / (C + D*sqrt(n))


try:
    complicated_solution = solve_ansatz(functional_equation, ComplicatedAnsatz(), n)
    display(complicated_solution)
except:
    print('Could not solve for complicated ansatz')
```

## Solving by plugging concrete values

```python
complicated_ansatz = ComplicatedAnsatz()

concrete_values = [1, 2]
equations = [
    functional_equation(complicated_ansatz, val)
    for val in concrete_values
]

args = sym.solve(equations, *complicated_ansatz.args, dict=True)

solution = complicated_ansatz(n).subs(args)
solution
```
